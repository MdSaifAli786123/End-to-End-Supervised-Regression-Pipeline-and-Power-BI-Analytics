# **STEP 5: Missing value Treatments**



*   Choose the method of inputting missing value- mean , median or mode
*   We can also use the algorithm to fill missing value of numerical column

*   For real or continous data: mean or median
*   For categorical or discrete data: mode

***For target variable we dont do any missing value and outlier treatment***





```python
df.isnull().sum()
```

```python
#fill missing value with median
for i in ['Diphtheria ','Polio','Income composition of resources']:
  df[i].fillna(df[i].median(),inplace=True)
```

```python
df.isnull().sum()
```

```python
#fill missing value with mode
for i in ["Adult Mortality"]:
  df[i].fillna(df[i].mode(),inplace=True)
```

```python
df.isnull().sum()
```

```python
#filling missing value using KNNImputer
from sklearn.impute import KNNImputer
impute=KNNImputer()
for i in df.select_dtypes(include="number").columns:
  df[i]=impute.fit_transform(df[[i]])
```

```python
df.isnull().sum()
```

# **STEP 6:Outlier Treatment**

1.   Its only done for continous data
2.   Not for categorical and discrete data

**Rule**: ***Make outlier beyond lower whisker has the value of lower whisker and outlier beyond upper whisker has the value of upper whisker***



```python
#doing outlier treatment
#not necessary we have to do for all
def whisker(col):
  q1,q3=np.percentile(col,[25,75])
  lqr=q3-q1
  upper_whisker=q3+1.5*lqr
  lower_whisker=q1-1.5*lqr
  return upper_whisker,lower_whisker
```

```python
df.columns
```

```python
#checking whisker function work or not
whisker(df['GDP'])
```

```python
#Treating the outlier of the below selected attributes
for i in ['GDP','Total expenditure',' thinness  1-19 years', ' thinness 5-9 years']:
  uw,lw=whisker(df[i])
  df[i]=np.where(df[i]<lw,lw,df[i])#read it as if..else like if an outlier greater then take uw else original value only
  df[i]=np.where(df[i]>uw,uw,df[i])#read it as if..else like if an outlier smaller then take lw else original value only
```

```python
#Checking whether outlier has been removed or not
for i in ['GDP','Total expenditure',' thinness  1-19 years', ' thinness 5-9 years']:
  sns.boxplot(df[i])
  plt.show()
#We see no outlier in the output boxplot
```

# **STEP 7:Duplicate and Garbage value treatment**

```python
#delete the duplicates value
df.drop_duplicates()
```

```python
#treatment of the garbage value with median and mode
# Step 1: Replace common garbage values with NaN
garbage_values = ["?", "N/A", "na", "null", "", "None", "-", "--"]
df.replace(garbage_values, np.nan, inplace=True)

# Step 2: Handle numerical columns (replace NaN with mean)
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    df[col].fillna(df[col].mean(), inplace=True)

# Step 3: Handle categorical columns (replace NaN with mode)
categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col].fillna(df[col].mode(), inplace=True)
```

# **STEP 8:Normalizaion**

*   We dont normalize target variable while learning
*   We normalize target variable while optmization



```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Initialize scalers
minmax_scaler = MinMaxScaler()
standard_scaler = StandardScaler()

# --------------------------------------------------
# Z-Score Normalization (for continuous features)
# --------------------------------------------------
# NOTE: Target column "Life expectancy " is EXCLUDED

zscore_columns = [
    "Alcohol",
    "Hepatitis B",
    "Total expenditure",
    "GDP"
]

for col in zscore_columns:
    df[col] = standard_scaler.fit_transform(df[[col]])

# --------------------------------------------------
# Min–Max Normalization (0 to 1)
# --------------------------------------------------

minmax_columns = [
    "Adult Mortality",
    "Schooling",
    "Population"
]

for col in minmax_columns:
    df[col] = minmax_scaler.fit_transform(df[[col]])

# Display normalized dataset
df.head()
```
